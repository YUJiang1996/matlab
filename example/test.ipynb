用于Matlab的Simulink模型绘图

```python
import matlab.engine
import matplotlib.pyplot as plt
eng = matlab.engine.start_matlab()
eng.cd(r'E:\example')
eng.load_system('test')
eng.set_param('test','StopTime','20',nargout=0)
eng.set_param('test/Sine','Amplitude','2',nargout=0)
eng.sim('test',nargout=0)
x=eng.workspace['tout']
y1=eng.workspace['sin']
y2=eng.workspace['sin1']
y3=eng.workspace['sin2']
y4=eng.workspace['sin3']
[plt.plot(x,y,label=l,lw=2) for y ,l in zip([y1,y2,y3,y4],['sin','sin1','sin2','sin3'])]
plt.xlabel('Time')
plt.xlim(0,20)
plt.ylabel('Amplitude')
plt.title('Sine Wave')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
eng.quit()
```
